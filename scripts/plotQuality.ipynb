# Show ranking tables of simulations against experimental data

This notebook shows different ranking views of simulations against experimental data.
It reads ranking tables from `FMDL_DATA_PATH/Ranking/`.

With the current FAIRMD tools, generate/update ranking files with:
`fmdl_match_experiments`
`fmdl_evaluate_quality`
`fmdl_make_ranking`

# Initialize FAIRMD Lipids databank

```python
import os
import sys
import subprocess

# This installs FAIRMD Lipids databank tools on Colab and initializes data.
if "google.colab" in sys.modules:
    subprocess.run([sys.executable, "-m", "pip", "install", "fairmd-lipids"], check=True)
    subprocess.run(["fmdl_initialize_data", "stable"], check=True)

    from pathlib import Path
    import urllib.request

    plot_tools_path = Path("plot_tools.py")
    if not plot_tools_path.exists():
        urllib.request.urlretrieve(
            "https://raw.githubusercontent.com/NMRLipids/databank-template/main/scripts/plot_tools.py",
            plot_tools_path,
        )

    # Load environment variables from databank_env.rc into this Python process.
    with open("databank_env.rc") as env_file:
        for line in env_file:
            line = line.strip()
            if line.startswith("export ") and "=" in line:
                key, value = line[len("export "):].split("=", 1)
                os.environ[key] = value

# On local machine, set FMDL_DATA_PATH to the initialized databank folder.
# Optionally set FMDL_SIMU_PATH if your simulations are in a non-default location.
# os.environ["FMDL_DATA_PATH"] = "/path/to/your/BilayerData"
```

```python
# Core FAIRMD Lipids routines and Ranking data path
import pandas as pd
from IPython.display import display, Markdown
from fairmd.lipids import FMDL_DATA_PATH
from fairmd.lipids.core import initialize_databank

systems = initialize_databank()
ranking_dir = os.path.join(FMDL_DATA_PATH, "Ranking")

def show_table(df: pd.DataFrame, sort_key: str, title: str, top_n: int = 20):
    if sort_key not in df.columns:
        print(f"Column '{sort_key}' not found for {title}.")
        return
    out = df.dropna(subset=[sort_key]).sort_values(by=sort_key, ascending=False).head(top_n)
    display(Markdown(f"### {title}"))
    display(out)
```

```python
from fairmd.lipids.api import get_quality
import numpy as np

# Optional: build a live quality table directly from systems (no ranking files needed).
rows = []
for system in systems:
    rows.append({
        "ID": system["ID"],
        "FF": system.get("FF", np.nan),
        "T": system.get("TEMPERATURE", np.nan),
        "Q_FF": get_quality(system, experiment="FF"),
        "Q_OP": get_quality(system, experiment="OP"),
        "Q_OP_tails": get_quality(system, part="tails", experiment="OP"),
        "Q_OP_heads": get_quality(system, part="headgroup", experiment="OP"),
    })

live_quality_df = pd.DataFrame(rows)
show_table(live_quality_df, "Q_OP", "Live OP ranking from get_quality()")
```

# New FAIRMD quality tools (current version)

In addition to reading CSV rankings, FAIRMD now provides built-in tools to recompute and inspect quality metrics:

- CLI pipeline: `fmdl_match_experiments` -> `fmdl_evaluate_quality` -> `fmdl_make_ranking`
- Programmatic API: `fairmd.lipids.api.get_quality(system, ...)` for FF/OP quality values

# Show ranking of simulations based on total quality, tails quality, and headgroup quality

```python
# Global rankings generated by fmdl_make_ranking
ranking_views = {
    "Form factor": ("FF_ranking.csv", "Q_FF"),
    "total": ("OP_ranking.csv", "Q_OP"),
    "tails": ("OP_ranking.csv", "Q_OP_tails"),
    "headgroup": ("OP_ranking.csv", "Q_OP_heads"),
}

for label, (filename, score_col) in ranking_views.items():
    path = os.path.join(ranking_dir, filename)
    if not os.path.isfile(path):
        print(f"Missing ranking file: {path}")
        continue
    df = pd.read_csv(path)
    show_table(df, score_col, f"Sorted by {label} quality")
```

# Show ranking separately for each lipid based on total, tails, or headgroup quality

```python
fragments = {
    "total": "Q_OP",
    "tails": "Q_OP_tails",
    "headgroup": "Q_OP_heads",
}

lipid_names = sorted({lip for system in systems for lip in system.lipids})

for fragment, score_col in fragments.items():
    for lipid_name in lipid_names:
        ranking_path = os.path.join(ranking_dir, f"{lipid_name}_ranking.csv")
        if not os.path.isfile(ranking_path):
            continue
        df = pd.read_csv(ranking_path)
        show_table(df, score_col, f"{lipid_name}: {fragment} quality", top_n=10)
```

# Show form factor ranking only for systems with a specific lipid (cholesterol as an example)

```python
# Ranking only for systems containing a specific lipid (cholesterol as an example)
lipid = "CHOL"
ranking_path = os.path.join(ranking_dir, "FF_ranking.csv")

df = pd.read_csv(ranking_path)
subset = df[df["composition"].str.contains(fr"\b{lipid}\b", regex=True, na=False)]
show_table(subset, "Q_FF", f"Form-factor ranking for systems with {lipid}", top_n=30)
```

# Show POPC tails ranking only for systems with POPC and CHOL

```python
# POPC ranking for systems containing cholesterol
lipid_filter = "CHOL"
ranking_path = os.path.join(ranking_dir, "POPC_ranking.csv")

df = pd.read_csv(ranking_path)
subset = df[df["composition"].str.contains(fr"\b{lipid_filter}\b", regex=True, na=False)]
show_table(subset, "Q_OP_tails", "POPC tails quality in systems with CHOL", top_n=30)
```
